### 1. Two Sum

Return indices of two values whose sum equals `target`.

[Open on LeetCode](https://leetcode.com/problems/two-sum/) · **Complexity:** O(n) time, O(n) space

```python
class Solution:
    def twoSum(self, nums: List[int], target: int) -> List[int]:
        seen = {}
        for i, x in enumerate(nums):
            if target - x in seen:
                return [seen[target - x], i]
            seen[x] = i
```

### 49. Group Anagrams

Group strings that are anagrams of one another.

[Open on LeetCode](https://leetcode.com/problems/group-anagrams/) · **Complexity:** O(nk log k) time, O(nk) space

```python
class Solution:
    def groupAnagrams(self, strs: List[str]) -> List[List[str]]:
        groups = defaultdict(list)
        for s in strs:
            groups[tuple(sorted(s))].append(s)
        return list(groups.values())
```

### 128. Longest Consecutive Sequence

Find the length of the longest run of consecutive integers in an unsorted array.

[Open on LeetCode](https://leetcode.com/problems/longest-consecutive-sequence/) · **Complexity:** O(n) time, O(n) space

```python
class Solution:
    def longestConsecutive(self, nums: List[int]) -> int:
        values, best = set(nums), 0
        for x in values:
            if x - 1 not in values:
                y = x
                while y in values:
                    y += 1
                best = max(best, y - x)
        return best
```

### 53. Maximum Subarray

Return the largest sum of a non-empty contiguous subarray.

[Open on LeetCode](https://leetcode.com/problems/maximum-subarray/) · **Complexity:** O(n) time, O(1) space

```python
class Solution:
    def maxSubArray(self, nums: List[int]) -> int:
        best = cur = nums[0]
        for x in nums[1:]:
            cur = max(x, cur + x)
            best = max(best, cur)
        return best
```

### 56. Merge Intervals

Merge all overlapping intervals.

[Open on LeetCode](https://leetcode.com/problems/merge-intervals/) · **Complexity:** O(n log n) time, O(n) output space

```python
class Solution:
    def merge(self, intervals: List[List[int]]) -> List[List[int]]:
        intervals.sort()
        out = []
        for start, end in intervals:
            if out and start <= out[-1][1]:
                out[-1][1] = max(out[-1][1], end)
            else:
                out.append([start, end])
        return out
```

### 189. Rotate Array

Rotate the array right by `k` positions in-place.

[Open on LeetCode](https://leetcode.com/problems/rotate-array/) · **Complexity:** O(n) time, O(1) space

```python
class Solution:
    def rotate(self, nums: List[int], k: int) -> None:
        k %= len(nums)

        def reverse(left, right):
            while left < right:
                nums[left], nums[right] = nums[right], nums[left]
                left += 1
                right -= 1

        reverse(0, len(nums) - 1)
        reverse(0, k - 1)
        reverse(k, len(nums) - 1)
```

### 238. Product of Array Except Self

For every index, return the product of all other elements without division.

[Open on LeetCode](https://leetcode.com/problems/product-of-array-except-self/) · **Complexity:** O(n) time, O(1) extra space excluding output

```python
class Solution:
    def productExceptSelf(self, nums: List[int]) -> List[int]:
        out, prefix = [1] * len(nums), 1
        for i, x in enumerate(nums):
            out[i], prefix = prefix, prefix * x
        suffix = 1
        for i in range(len(nums) - 1, -1, -1):
            out[i], suffix = out[i] * suffix, suffix * nums[i]
        return out
```

### 41. First Missing Positive

Find the smallest missing positive integer using constant extra space.

[Open on LeetCode](https://leetcode.com/problems/first-missing-positive/) · **Complexity:** O(n) time, O(1) space

```python
class Solution:
    def firstMissingPositive(self, nums: List[int]) -> int:
        n = len(nums)
        for i in range(n):
            while 1 <= nums[i] <= n and nums[nums[i] - 1] != nums[i]:
                j = nums[i] - 1
                nums[i], nums[j] = nums[j], nums[i]
        for i, x in enumerate(nums):
            if x != i + 1:
                return i + 1
        return n + 1
```

### 136. Single Number

Every value appears twice except one; return the unique value.

[Open on LeetCode](https://leetcode.com/problems/single-number/) · **Complexity:** O(n) time, O(1) space

```python
class Solution:
    def singleNumber(self, nums: List[int]) -> int:
        ans = 0
        for x in nums:
            ans ^= x
        return ans
```

### 169. Majority Element

Return the element appearing more than half the time.

[Open on LeetCode](https://leetcode.com/problems/majority-element/) · **Complexity:** O(n) time, O(1) space

```python
class Solution:
    def majorityElement(self, nums: List[int]) -> int:
        candidate, votes = None, 0
        for x in nums:
            if votes == 0:
                candidate = x
            votes += 1 if x == candidate else -1
        return candidate
```

### 75. Sort Colors

In-place sort an array containing only 0, 1, and 2.

[Open on LeetCode](https://leetcode.com/problems/sort-colors/) · **Complexity:** O(n) time, O(1) space

```python
class Solution:
    def sortColors(self, nums: List[int]) -> None:
        low = i = 0
        high = len(nums) - 1
        while i <= high:
            if nums[i] == 0:
                nums[low], nums[i] = nums[i], nums[low]
                low += 1
                i += 1
            elif nums[i] == 2:
                nums[high], nums[i] = nums[i], nums[high]
                high -= 1
            else:
                i += 1
```

### 31. Next Permutation

Transform the array into its next lexicographically greater permutation in-place.

[Open on LeetCode](https://leetcode.com/problems/next-permutation/) · **Complexity:** O(n) time, O(1) space

```python
class Solution:
    def nextPermutation(self, nums: List[int]) -> None:
        i = len(nums) - 2
        while i >= 0 and nums[i] >= nums[i + 1]:
            i -= 1
        if i >= 0:
            j = len(nums) - 1
            while nums[j] <= nums[i]:
                j -= 1
            nums[i], nums[j] = nums[j], nums[i]
        nums[i + 1 :] = reversed(nums[i + 1 :])
```

### 287. Find the Duplicate Number

Find the repeated value without modifying the array and with constant extra space.

[Open on LeetCode](https://leetcode.com/problems/find-the-duplicate-number/) · **Complexity:** O(n) time, O(1) space

```python
class Solution:
    def findDuplicate(self, nums: List[int]) -> int:
        slow = fast = nums[0]
        while True:
            slow, fast = nums[slow], nums[nums[fast]]
            if slow == fast:
                break
        slow = nums[0]
        while slow != fast:
            slow, fast = nums[slow], nums[fast]
        return slow
```

### 283. Move Zeroes

```python
class Solution:
    def moveZeroes(self, nums: List[int]) -> None:
        w = 0
        for x in nums:
            if x != 0:
                nums[w] = x
                w += 1
        nums[w:] = [0] * (len(nums) - w)
```

### 11. Container With Most Water

Choose two vertical lines that contain the most water.

[Open on LeetCode](https://leetcode.com/problems/container-with-most-water/) · **Complexity:** O(n) time, O(1) space

```python
class Solution:
    def maxArea(self, height: List[int]) -> int:
        l, r, best = 0, len(height) - 1, 0
        while l < r:
            best = max(best, (r - l) * min(height[l], height[r]))
            if height[l] < height[r]:
                l += 1
            else:
                r -= 1
        return best
```

### 15. 3Sum

Return all unique triplets whose sum is zero.

[Open on LeetCode](https://leetcode.com/problems/3sum/) · **Complexity:** O(n²) time, O(1) extra space excluding output

```python
class Solution:
    def threeSum(self, nums: List[int]) -> List[List[int]]:
        nums.sort()
        out = []
        for i in range(len(nums) - 2):
            if i and nums[i] == nums[i - 1]:
                continue
            l, r = i + 1, len(nums) - 1
            while l < r:
                s = nums[i] + nums[l] + nums[r]
                if s < 0:
                    l += 1
                elif s > 0:
                    r -= 1
                else:
                    out.append([nums[i], nums[l], nums[r]])
                    l += 1
                    r -= 1
                    while l < r and nums[l] == nums[l - 1]:
                        l += 1
        return out
```

### 42. Trapping Rain Water

Compute how much rainwater is trapped between elevation bars.

[Open on LeetCode](https://leetcode.com/problems/trapping-rain-water/) · **Complexity:** O(n) time, O(1) space

```python
class Solution:
    def trap(self, height: List[int]) -> int:
        l, r, lm, rm, water = 0, len(height) - 1, 0, 0, 0
        while l < r:
            if height[l] < height[r]:
                lm = max(lm, height[l])
                water += lm - height[l]
                l += 1
            else:
                rm = max(rm, height[r])
                water += rm - height[r]
                r -= 1
        return water
```

### 3. Longest Substring Without Repeating Characters

Find the longest substring containing no repeated characters.

[Open on LeetCode](https://leetcode.com/problems/longest-substring-without-repeating-characters/) · **Complexity:** O(n) time, O(k) space

```python
class Solution:
    def lengthOfLongestSubstring(self, s: str) -> int:
        last, left, best = {}, 0, 0
        for right, ch in enumerate(s):
            left = max(left, last.get(ch, -1) + 1)
            last[ch] = right
            best = max(best, right - left + 1)
        return best
```

### 438. Find All Anagrams in a String

Return start indices of substrings of `s` that are anagrams of `p`.

[Open on LeetCode](https://leetcode.com/problems/find-all-anagrams-in-a-string/) · **Complexity:** O(n) time, O(k) space

```python
class Solution:
    def findAnagrams(self, s: str, p: str) -> List[int]:
        need, window, out, k = Counter(p), Counter(), [], len(p)
        for i, ch in enumerate(s):
            window[ch] += 1
            if i >= k:
                old = s[i - k]
                window[old] -= 1
                if window[old] == 0:
                    del window[old]
            if window == need:
                out.append(i - k + 1)
        return out
```

### 560. Subarray Sum Equals K

Count contiguous subarrays whose sum equals `k`.

[Open on LeetCode](https://leetcode.com/problems/subarray-sum-equals-k/) · **Complexity:** O(n) time, O(n) space

```python
class Solution:
    def subarraySum(self, nums: List[int], k: int) -> int:
        freq, prefix, ans = {0: 1}, 0, 0
        for x in nums:
            prefix += x
            ans += freq.get(prefix - k, 0)
            freq[prefix] = freq.get(prefix, 0) + 1
        return ans
```

### 239. Sliding Window Maximum

Return the maximum value in every window of size `k`.

[Open on LeetCode](https://leetcode.com/problems/sliding-window-maximum/) · **Complexity:** O(n) time, O(k) space

```python
class Solution:
    def maxSlidingWindow(self, nums: List[int], k: int) -> List[int]:
        q, out = deque(), []
        for i, x in enumerate(nums):
            while q and nums[q[-1]] <= x:
                q.pop()
            q.append(i)
            if q[0] <= i - k:
                q.popleft()
            if i >= k - 1:
                out.append(nums[q[0]])
        return out
```

### 76. Minimum Window Substring

Find the shortest substring of `s` containing every character of `t` with multiplicity.

[Open on LeetCode](https://leetcode.com/problems/minimum-window-substring/) · **Complexity:** O(n) time, O(k) space

```python
class Solution:
    def minWindow(self, s: str, t: str) -> str:
        need, missing, left, start, length = Counter(t), len(t), 0, 0, inf
        for right, ch in enumerate(s):
            if need[ch] > 0:
                missing -= 1
            need[ch] -= 1
            while missing == 0:
                if right - left + 1 < length:
                    start, length = left, right - left + 1
                need[s[left]] += 1
                if need[s[left]] > 0:
                    missing += 1
                left += 1
        return "" if length == inf else s[start : start + length]
```

### 73. Set Matrix Zeroes

If a cell is zero, set its entire row and column to zero in-place.

[Open on LeetCode](https://leetcode.com/problems/set-matrix-zeroes/) · **Complexity:** O(mn) time, O(1) space

```python
class Solution:
    def setZeroes(self, matrix: List[List[int]]) -> None:
        first_col = any(row[0] == 0 for row in matrix)
        first_row = any(x == 0 for x in matrix[0])
        for i in range(1, len(matrix)):
            for j in range(1, len(matrix[0])):
                if matrix[i][j] == 0:
                    matrix[i][0] = matrix[0][j] = 0
        for i in range(1, len(matrix)):
            for j in range(1, len(matrix[0])):
                if matrix[i][0] == 0 or matrix[0][j] == 0:
                    matrix[i][j] = 0
        if first_row:
            matrix[0] = [0] * len(matrix[0])
        if first_col:
            for row in matrix:
                row[0] = 0
```

### 54. Spiral Matrix

Return matrix elements in clockwise spiral order.

[Open on LeetCode](https://leetcode.com/problems/spiral-matrix/) · **Complexity:** O(mn) time, O(1) extra space

```python
class Solution:
    def spiralOrder(self, matrix: List[List[int]]) -> List[int]:
        out = []
        while matrix:
            out += matrix.pop(0)
            if matrix and matrix[0]:
                for row in matrix:
                    out.append(row.pop())
            if matrix:
                out += matrix.pop()[::-1]
            if matrix and matrix[0]:
                for row in matrix[::-1]:
                    out.append(row.pop(0))
        return out
```

### 240. Search a 2D Matrix II

Search a matrix whose rows and columns are each sorted ascending.

```python
class Solution:
    def searchMatrix(self, matrix: List[List[int]], target: int) -> bool:
        r, c = 0, len(matrix[0]) - 1
        while r < len(matrix) and c >= 0:
            if matrix[r][c] == target:
                return True
            if matrix[r][c] > target:
                c -= 1
            else:
                r += 1
        return False
```

### 74. Search a 2D Matrix

Search a row-major globally sorted matrix.

[Open on LeetCode](https://leetcode.com/problems/search-a-2d-matrix/) · **Complexity:** O(log(mn)) time, O(1) space

```python
class Solution:
    def searchMatrix(self, matrix: List[List[int]], target: int) -> bool:
        m, n = len(matrix), len(matrix[0])
        l, r = 0, m * n - 1
        while l <= r:
            mid = (l + r) // 2
            x = matrix[mid // n][mid % n]
            if x == target:
                return True
            if x < target:
                l = mid + 1
            else:
                r = mid - 1
        return False
```

### 34. Find First and Last Position of Element in Sorted Array

Return the first and last index of `target` in a sorted array.

[Open on LeetCode](https://leetcode.com/problems/find-first-and-last-position-of-element-in-sorted-array/) · **Complexity:** O(log n) time, O(1) space

```python
class Solution:
    def searchRange(self, nums: List[int], target: int) -> List[int]:
        left = bisect_left(nums, target)
        right = bisect_left(nums, target + 1) - 1
        return [left, right] if left < len(nums) and nums[left] == target else [-1, -1]
```

### 33. Search in Rotated Sorted Array

Find `target` in a distinct sorted array rotated at an unknown pivot.

[Open on LeetCode](https://leetcode.com/problems/search-in-rotated-sorted-array/) · **Complexity:** O(log n) time, O(1) space

```python
class Solution:
    def search(self, nums: List[int], target: int) -> int:
        l, r = 0, len(nums) - 1
        while l <= r:
            m = (l + r) // 2
            if nums[m] == target:
                return m
            if nums[l] <= nums[m]:
                if nums[l] <= target < nums[m]:
                    r = m - 1
                else:
                    l = m + 1
            elif nums[m] < target <= nums[r]:
                l = m + 1
            else:
                r = m - 1
        return -1
```

### 153. Find Minimum in Rotated Sorted Array

Return the minimum of a rotated array of distinct values.

[Open on LeetCode](https://leetcode.com/problems/find-minimum-in-rotated-sorted-array/) · **Complexity:** O(log n) time, O(1) space

```python
class Solution:
    def findMin(self, nums: List[int]) -> int:
        l, r = 0, len(nums) - 1
        while l < r:
            m = (l + r) // 2
            if nums[m] > nums[r]:
                l = m + 1
            else:
                r = m
        return nums[l]
```

### 4. Median of Two Sorted Arrays

Find the median of two sorted arrays in logarithmic time.

[Open on LeetCode](https://leetcode.com/problems/median-of-two-sorted-arrays/) · **Complexity:** O(log(min(m,n))) time, O(1) space

```python
class Solution:
    def findMedianSortedArrays(self, a: List[int], b: List[int]) -> float:
        if len(a) > len(b):
            a, b = b, a
        total, half = len(a) + len(b), (len(a) + len(b) + 1) // 2
        l, r = 0, len(a)
        while l <= r:
            i = (l + r) // 2
            j = half - i
            al = a[i - 1] if i else -inf
            ar = a[i] if i < len(a) else inf
            bl = b[j - 1] if j else -inf
            br = b[j] if j < len(b) else inf
            if al <= br and bl <= ar:
                return max(al, bl) if total % 2 else (max(al, bl) + min(ar, br)) / 2
            if al > br:
                r = i - 1
            else:
                l = i + 1
```

### 160. Intersection of Two Linked Lists

Return the node where two singly linked lists intersect, or `None`.

[Open on LeetCode](https://leetcode.com/problems/intersection-of-two-linked-lists/) · **Complexity:** O(m+n) time, O(1) space

```python
class Solution:
    def getIntersectionNode(
        self, headA: ListNode, headB: ListNode
    ) -> Optional[ListNode]:
        a, b = headA, headB
        while a is not b:
            a = a.next if a else headB
            b = b.next if b else headA
        return a
```

### 206. Reverse Linked List

Reverse a singly linked list.

```python
class Solution:
    def reverseList(self, head: Optional[ListNode]) -> Optional[ListNode]:
        prev = None
        while head:
            head.next, prev, head = prev, head, head.next
        return prev
```

### 234. Palindrome Linked List

Determine whether a linked list reads the same forward and backward.

[Open on LeetCode](https://leetcode.com/problems/palindrome-linked-list/) · **Complexity:** O(n) time, O(1) space

```python
class Solution:
    def isPalindrome(self, head: Optional[ListNode]) -> bool:
        slow = fast = head
        while fast and fast.next:
            slow, fast = slow.next, fast.next.next
        prev = None
        while slow:
            slow.next, prev, slow = prev, slow, slow.next
        while prev:
            if head.val != prev.val:
                return False
            head, prev = head.next, prev.next
        return True
```

### 141. Linked List Cycle

Return whether a linked list contains a cycle.

[Open on LeetCode](https://leetcode.com/problems/linked-list-cycle/) · **Complexity:** O(n) time, O(1) space

```python
class Solution:
    def hasCycle(self, head: Optional[ListNode]) -> bool:
        slow = fast = head
        while fast and fast.next:
            slow, fast = slow.next, fast.next.next
            if slow is fast:
                return True
        return False
```

### 142. Linked List Cycle II

Return the node where a cycle begins, or `None`.

[Open on LeetCode](https://leetcode.com/problems/linked-list-cycle-ii/) · **Complexity:** O(n) time, O(1) space

```python
class Solution:
    def detectCycle(self, head: Optional[ListNode]) -> Optional[ListNode]:
        slow = fast = head
        while fast and fast.next:
            slow, fast = slow.next, fast.next.next
            if slow is fast:
                slow = head
                while slow is not fast:
                    slow, fast = slow.next, fast.next
                return slow
        return None
```

### 21. Merge Two Sorted Lists

Merge two sorted linked lists.

[Open on LeetCode](https://leetcode.com/problems/merge-two-sorted-lists/) · **Complexity:** O(m+n) time, O(1) space

```python
class Solution:
    def mergeTwoLists(
        self, a: Optional[ListNode], b: Optional[ListNode]
    ) -> Optional[ListNode]:
        dummy = tail = ListNode()
        while a and b:
            if a.val < b.val:
                tail.next, a = a, a.next
            else:
                tail.next, b = b, b.next
            tail = tail.next
        tail.next = a or b
        return dummy.next
```

### 2. Add Two Numbers

Add two reverse-order digit lists and return the sum as a list.

[Open on LeetCode](https://leetcode.com/problems/add-two-numbers/) · **Complexity:** O(max(m,n)) time, O(max(m,n)) output space

```python
class Solution:
    def addTwoNumbers(
        self, a: Optional[ListNode], b: Optional[ListNode]
    ) -> Optional[ListNode]:
        dummy = tail = ListNode()
        carry = 0
        while a or b or carry:
            total = (a.val if a else 0) + (b.val if b else 0) + carry
            carry, digit = divmod(total, 10)
            tail.next = ListNode(digit)
            tail = tail.next
            a = a.next if a else None
            b = b.next if b else None
        return dummy.next
```

### 19. Remove Nth Node From End of List

Remove the nth node from the end in one pass.

[Open on LeetCode](https://leetcode.com/problems/remove-nth-node-from-end-of-list/) · **Complexity:** O(n) time, O(1) space

```python
class Solution:
    def removeNthFromEnd(self, head: Optional[ListNode], n: int) -> Optional[ListNode]:
        dummy = ListNode(0, head)
        fast = slow = dummy
        for _ in range(n):
            fast = fast.next
        while fast.next:
            fast, slow = fast.next, slow.next
        slow.next = slow.next.next
        return dummy.next
```

### 24. Swap Nodes in Pairs

Swap every adjacent pair of linked-list nodes.

[Open on LeetCode](https://leetcode.com/problems/swap-nodes-in-pairs/) · **Complexity:** O(n) time, O(1) space

```python
class Solution:
    def swapPairs(self, head: Optional[ListNode]) -> Optional[ListNode]:
        dummy = ListNode(0, head)
        prev = dummy
        while prev.next and prev.next.next:
            a, b = prev.next, prev.next.next
            prev.next, a.next, b.next = b, b.next, a
            prev = a
        return dummy.next
```

### 25. Reverse Nodes in k-Group

Reverse list nodes in groups of `k`; leave an incomplete final group unchanged.

[Open on LeetCode](https://leetcode.com/problems/reverse-nodes-in-k-group/) · **Complexity:** O(n) time, O(1) space

```python
class Solution:
    def reverseKGroup(self, head: Optional[ListNode], k: int) -> Optional[ListNode]:
        dummy = ListNode(0, head)
        group_prev = dummy
        while True:
            kth = group_prev
            for _ in range(k):
                kth = kth.next
                if not kth:
                    return dummy.next
            group_next, prev, cur = kth.next, kth.next, group_prev.next
            while cur is not group_next:
                cur.next, prev, cur = prev, cur, cur.next
            old_start = group_prev.next
            group_prev.next = kth
            group_prev = old_start
```

### 138. Copy List with Random Pointer

Deep-copy a linked list whose nodes also have random pointers.

[Open on LeetCode](https://leetcode.com/problems/copy-list-with-random-pointer/) · **Complexity:** O(n) time, O(n) space

```python
class Solution:
    def copyRandomList(self, head: Optional[Node]) -> Optional[Node]:
        copies = {None: None}
        cur = head
        while cur:
            copies[cur] = Node(cur.val)
            cur = cur.next
        cur = head
        while cur:
            copies[cur].next, copies[cur].random = copies[cur.next], copies[cur.random]
            cur = cur.next
        return copies[head]
```

### 148. Sort List

Sort a linked list in ascending order.

[Open on LeetCode](https://leetcode.com/problems/sort-list/) · **Complexity:** O(n log n) time, O(log n) recursion space

```python
class Solution:
    def sortList(self, head: Optional[ListNode]) -> Optional[ListNode]:
        if not head or not head.next:
            return head
        slow, fast = head, head.next
        while fast and fast.next:
            slow, fast = slow.next, fast.next.next
        right = slow.next
        slow.next = None
        a, b = self.sortList(head), self.sortList(right)
        dummy = tail = ListNode()
        while a and b:
            if a.val < b.val:
                tail.next, a = a, a.next
            else:
                tail.next, b = b, b.next
            tail = tail.next
        tail.next = a or b
        return dummy.next
```

### 23. Merge k Sorted Lists

Merge `k` sorted linked lists into one sorted list.

[Open on LeetCode](https://leetcode.com/problems/merge-k-sorted-lists/) · **Complexity:** O(N log k) time, O(k) space

```python
class Solution:
    def mergeKLists(self, lists: List[Optional[ListNode]]) -> Optional[ListNode]:
        heap = []
        for i, node in enumerate(lists):
            if node:
                heappush(heap, (node.val, i, node))
        dummy = tail = ListNode()
        while heap:
            _, i, node = heappop(heap)
            tail.next = node
            tail = node
            if node.next:
                heappush(heap, (node.next.val, i, node.next))
        return dummy.next
```

### 146. LRU Cache

Design an O(1) least-recently-used cache with `get` and `put`.

[Open on LeetCode](https://leetcode.com/problems/lru-cache/) · **Complexity:** O(1) per operation, O(capacity) space

```python
class LRUCache:
    def __init__(self, capacity: int):
        self.capacity, self.data = capacity, OrderedDict()

    def get(self, key: int) -> int:
        if key not in self.data:
            return -1
        self.data.move_to_end(key)
        return self.data[key]

    def put(self, key: int, value: int) -> None:
        self.data[key] = value
        self.data.move_to_end(key)
        if len(self.data) > self.capacity:
            self.data.popitem(last=False)
```

Design a stack supporting push, pop, top, and minimum in O(1).

[Open on LeetCode](https://leetcode.com/problems/min-stack/) · **Complexity:** O(1) per operation, O(n) space

```python
class MinStack:
    def __init__(self):
        self.stack = []

    def push(self, val: int) -> None:
        self.stack.append((val, min(val, self.stack[-1][1]) if self.stack else val))

    def pop(self) -> None:
        self.stack.pop()

    def top(self) -> int:
        return self.stack[-1][0]

    def getMin(self) -> int:
        return self.stack[-1][1]
```

### 394. Decode String

Decode nested repetition expressions such as `3[a2[c]]`.

[Open on LeetCode](https://leetcode.com/problems/decode-string/) · **Complexity:** O(output length) time, O(n) space

```python
class Solution:
    def decodeString(self, s: str) -> str:
        stack, cur, num = [], "", 0
        for ch in s:
            if ch.isdigit():
                num = num * 10 + int(ch)
            elif ch == "[":
                stack.append((cur, num))
                cur, num = "", 0
            elif ch == "]":
                prev, count = stack.pop()
                cur = prev + count * cur
            else:
                cur += ch
        return cur
```

```python
class Solution:
    def largestRectangleArea(self, heights: List[int]) -> int:
        stack, best = [], 0
        for i, h in enumerate(heights + [0]):
            start = i
            while stack and stack[-1][1] > h:
                start, old = stack.pop()
                best = max(best, old * (i - start))
            stack.append((start, h))
        return best
```

Maintain a stream of numbers and report its median.

[Open on LeetCode](https://leetcode.com/problems/find-median-from-data-stream/) · **Complexity:** O(log n) add, O(1) median, O(n) space

```python
class MedianFinder:
    def __init__(self):
        self.low, self.high = [], []

    def addNum(self, num: int) -> None:
        heappush(self.low, -num)
        heappush(self.high, -heappop(self.low))
        if len(self.high) > len(self.low):
            heappush(self.low, -heappop(self.high))

    def findMedian(self) -> float:
        return (
            -self.low[0]
            if len(self.low) > len(self.high)
            else (-self.low[0] + self.high[0]) / 2
        )
```

Return the longest path between any two tree nodes, measured in edges.

```python
class Solution:
    def diameterOfBinaryTree(self, root: Optional[TreeNode]) -> int:
        best = 0

        def depth(node):
            nonlocal best
            if not node:
                return 0
            l, r = depth(node.left), depth(node.right)
            best = max(best, l + r)
            return 1 + max(l, r)

        depth(root)
        return best
```

### 102. Binary Tree Level Order Traversal

Return node values grouped by depth.

[Open on LeetCode](https://leetcode.com/problems/binary-tree-level-order-traversal/) · **Complexity:** O(n) time, O(n) space

```python
class Solution:
    def levelOrder(self, root: Optional[TreeNode]) -> List[List[int]]:
        if not root:
            return []
        q, out = deque([root]), []
        while q:
            level = []
            for _ in range(len(q)):
                node = q.popleft()
                level.append(node.val)
                if node.left:
                    q.append(node.left)
                if node.right:
                    q.append(node.right)
            out.append(level)
        return out
```

```python
class Solution:
    def sortedArrayToBST(self, nums: List[int]) -> Optional[TreeNode]:
        def build(left, right):
            if left > right:
                return None
            mid = (left + right) // 2
            return TreeNode(nums[mid], build(left, mid - 1), build(mid + 1, right))

        return build(0, len(nums) - 1)
```

### 437. Path Sum III

Count downward tree paths whose values sum to `targetSum`.

[Open on LeetCode](https://leetcode.com/problems/path-sum-iii/) · **Complexity:** O(n) time, O(h) space

```python
class Solution:
    def pathSum(self, root: Optional[TreeNode], targetSum: int) -> int:
        freq = {0: 1}

        def dfs(node, total):
            if not node:
                return 0
            total += node.val
            ans = freq.get(total - targetSum, 0)
            freq[total] = freq.get(total, 0) + 1
            ans += dfs(node.left, total) + dfs(node.right, total)
            freq[total] -= 1
            return ans

        return dfs(root, 0)
```

### 124. Binary Tree Maximum Path Sum

[Open on LeetCode](https://leetcode.com/problems/binary-tree-maximum-path-sum/) · **Complexity:** O(n) time, O(h) space

```python
class Solution:
    def maxPathSum(self, root: Optional[TreeNode]) -> int:
        best = -inf

        def gain(node):
            nonlocal best
            if not node:
                return 0
            l, r = max(0, gain(node.left)), max(0, gain(node.right))
            best = max(best, node.val + l + r)
            return node.val + max(l, r)

        gain(root)
        return best
```

### 200. Number of Islands

Count connected components of land in a binary grid.

[Open on LeetCode](https://leetcode.com/problems/number-of-islands/) · **Complexity:** O(mn) time, O(mn) worst-case stack

```python
class Solution:
    def numIslands(self, grid: List[List[str]]) -> int:
        def sink(r, c):
            if (
                r < 0
                or c < 0
                or r == len(grid)
                or c == len(grid[0])
                or grid[r][c] != "1"
            ):
                return
            grid[r][c] = "0"
            for dr, dc in ((1, 0), (-1, 0), (0, 1), (0, -1)):
                sink(r + dr, c + dc)

        count = 0
        for r in range(len(grid)):
            for c in range(len(grid[0])):
                if grid[r][c] == "1":
                    count += 1
                    sink(r, c)
        return count
```

### 994. Rotting Oranges

Return minutes until all fresh oranges rot via four-directional spread, or -1.

[Open on LeetCode](https://leetcode.com/problems/rotting-oranges/) · **Complexity:** O(mn) time, O(mn) space

```python
class Solution:
    def orangesRotting(self, grid: List[List[int]]) -> int:
        q, fresh = deque(), 0
        for r, row in enumerate(grid):
            for c, x in enumerate(row):
                if x == 2:
                    q.append((r, c, 0))
                elif x == 1:
                    fresh += 1
        minutes = 0
        while q:
            r, c, minutes = q.popleft()
            for dr, dc in ((1, 0), (-1, 0), (0, 1), (0, -1)):
                nr, nc = r + dr, c + dc
                if 0 <= nr < len(grid) and 0 <= nc < len(grid[0]) and grid[nr][nc] == 1:
                    grid[nr][nc] = 2
                    fresh -= 1
                    q.append((nr, nc, minutes + 1))
        return minutes if fresh == 0 else -1
```

### 207. Course Schedule

Determine whether all courses can be completed given prerequisite edges.

[Open on LeetCode](https://leetcode.com/problems/course-schedule/) · **Complexity:** O(V+E) time, O(V+E) space

```python
class Solution:
    def canFinish(self, numCourses: int, prerequisites: List[List[int]]) -> bool:
        graph, indegree = [[] for _ in range(numCourses)], [0] * numCourses
        for course, pre in prerequisites:
            graph[pre].append(course)
            indegree[course] += 1
        q = deque(i for i, d in enumerate(indegree) if d == 0)
        taken = 0
        while q:
            u = q.popleft()
            taken += 1
            for v in graph[u]:
                indegree[v] -= 1
                if indegree[v] == 0:
                    q.append(v)
        return taken == numCourses
```

### 208. Implement Trie (Prefix Tree)

Implement insertion, exact search, and prefix search for lowercase words.

[Open on LeetCode](https://leetcode.com/problems/implement-trie-prefix-tree/) · **Complexity:** O(L) per operation, O(total characters) space

```python
class Trie:
    def __init__(self):
        self.root = {}

    def insert(self, word: str) -> None:
        node = self.root
        for ch in word:
            node = node.setdefault(ch, {})
        node["#"] = True

    def search(self, word: str) -> bool:
        node = self.root
        for ch in word:
            if ch not in node:
                return False
            node = node[ch]
        return "#" in node

    def startsWith(self, prefix: str) -> bool:
        node = self.root
        for ch in prefix:
            if ch not in node:
                return False
            node = node[ch]
        return True
```

### 46. Permutations

Return every permutation of distinct values.

[Open on LeetCode](https://leetcode.com/problems/permutations/) · **Complexity:** O(n·n!) time, O(n) auxiliary space

```python
class Solution:
    def permute(self, nums: List[int]) -> List[List[int]]:
        out = []

        def backtrack(i):
            if i == len(nums):
                out.append(nums[:])
                return
            for j in range(i, len(nums)):
                nums[i], nums[j] = nums[j], nums[i]
                backtrack(i + 1)
                nums[i], nums[j] = nums[j], nums[i]

        backtrack(0)
        return out
```

### 39. Combination Sum

Return combinations of reusable candidates that sum to `target`.

[Open on LeetCode](https://leetcode.com/problems/combination-sum/) · **Complexity:** Exponential time, O(target/min candidate) stack

```python
class Solution:
    def combinationSum(self, candidates: List[int], target: int) -> List[List[int]]:
        out = []

        def dfs(start, remain, path):
            if remain == 0:
                out.append(path[:])
                return
            for i in range(start, len(candidates)):
                x = candidates[i]
                if x <= remain:
                    path.append(x)
                    dfs(i, remain - x, path)
                    path.pop()

        dfs(0, target, [])
        return out
```

```python
class Solution:
    def generateParenthesis(self, n: int) -> List[str]:
        out = []

        def build(s, opened, closed):
            if len(s) == 2 * n:
                out.append(s)
                return
            if opened < n:
                build(s + "(", opened + 1, closed)
            if closed < opened:
                build(s + ")", opened, closed + 1)

        build("", 0, 0)
        return out
```

### 79. Word Search

Determine whether a word can be traced through adjacent grid cells without reuse.

[Open on LeetCode](https://leetcode.com/problems/word-search/) · **Complexity:** O(mn·4ᴸ) time, O(L) space

```python
class Solution:
    def exist(self, board: List[List[str]], word: str) -> bool:
        def dfs(r, c, i):
            if i == len(word):
                return True
            if (
                not (0 <= r < len(board) and 0 <= c < len(board[0]))
                or board[r][c] != word[i]
            ):
                return False
            ch = board[r][c]
            board[r][c] = "#"
            found = any(
                dfs(r + dr, c + dc, i + 1)
                for dr, dc in ((1, 0), (-1, 0), (0, 1), (0, -1))
            )
            board[r][c] = ch
            return found

        return any(
            dfs(r, c, 0) for r in range(len(board)) for c in range(len(board[0]))
        )
```

### 131. Palindrome Partitioning

Return all ways to split a string into palindromic substrings.

[Open on LeetCode](https://leetcode.com/problems/palindrome-partitioning/) · **Complexity:** O(n·2ⁿ) time, O(n) auxiliary space

```python
class Solution:
    def partition(self, s: str) -> List[List[str]]:
        out = []

        def dfs(i, path):
            if i == len(s):
                out.append(path[:])
                return
            for j in range(i, len(s)):
                part = s[i : j + 1]
                if part == part[::-1]:
                    path.append(part)
                    dfs(j + 1, path)
                    path.pop()

        dfs(0, [])
        return out
```

Return every placement of `n` queens with no attacks.

[Open on LeetCode](https://leetcode.com/problems/n-queens/) · **Complexity:** O(n!) time, O(n) auxiliary space

```python
class Solution:
    def solveNQueens(self, n: int) -> List[List[str]]:
        out = []
        cols = set()
        diag = set()
        anti = set()
        board = [["."] * n for _ in range(n)]

        def dfs(r):
            if r == n:
                out.append(["".join(row) for row in board])
                return
            for c in range(n):
                if c in cols or r - c in diag or r + c in anti:
                    continue
                cols.add(c)
                diag.add(r - c)
                anti.add(r + c)
                board[r][c] = "Q"
                dfs(r + 1)
                board[r][c] = "."
                cols.remove(c)
                diag.remove(r - c)
                anti.remove(r + c)

        dfs(0)
        return out
```

### 70. Climbing Stairs

Count ways to climb `n` steps using moves of one or two.

[Open on LeetCode](https://leetcode.com/problems/climbing-stairs/) · **Complexity:** O(n) time, O(1) space

```python
class Solution:
    def climbStairs(self, n: int) -> int:
        a, b = 1, 1
        for _ in range(n):
            a, b = b, a + b
        return a
```

### 118. Pascal's Triangle

Return the first `numRows` rows of Pascal's triangle.

[Open on LeetCode](https://leetcode.com/problems/pascals-triangle/) · **Complexity:** O(n²) time and output space

```python
class Solution:
    def generate(self, numRows: int) -> List[List[int]]:
        rows = []
        for i in range(numRows):
            row = [1] * (i + 1)
            for j in range(1, i):
                row[j] = rows[-1][j - 1] + rows[-1][j]
            rows.append(row)
        return rows
```

### 198. House Robber

Maximize loot without choosing adjacent houses.

[Open on LeetCode](https://leetcode.com/problems/house-robber/) · **Complexity:** O(n) time, O(1) space

```python
class Solution:
    def rob(self, nums: List[int]) -> int:
        prev, cur = 0, 0
        for x in nums:
            prev, cur = cur, max(cur, prev + x)
        return cur
```

### 279. Perfect Squares

Return the fewest perfect-square values summing to `n`.

[Open on LeetCode](https://leetcode.com/problems/perfect-squares/) · **Complexity:** O(n√n) time, O(n) space

```python
class Solution:
    def numSquares(self, n: int) -> int:
        dp = [0] + [inf] * n
        for i in range(1, n + 1):
            dp[i] = 1 + min(dp[i - j * j] for j in range(1, int(i**0.5) + 1))
        return dp[n]
```

### 322. Coin Change

Return the fewest coins needed for `amount`, or -1.

[Open on LeetCode](https://leetcode.com/problems/coin-change/) · **Complexity:** O(amount·coins) time, O(amount) space

```python
class Solution:
    def coinChange(self, coins: List[int], amount: int) -> int:
        dp = [0] + [amount + 1] * amount
        for value in range(1, amount + 1):
            dp[value] = min(
                (dp[value - c] + 1 for c in coins if c <= value), default=amount + 1
            )
        return -1 if dp[amount] > amount else dp[amount]
```

### 139. Word Break

Determine whether a string can be segmented into dictionary words.

[Open on LeetCode](https://leetcode.com/problems/word-break/) · **Complexity:** O(n²) time, O(n) space

```python
class Solution:
    def wordBreak(self, s: str, wordDict: List[str]) -> bool:
        words = set(wordDict)
        dp = [True] + [False] * len(s)
        for i in range(1, len(s) + 1):
            dp[i] = any(dp[j] and s[j:i] in words for j in range(i))
        return dp[-1]
```

### 300. Longest Increasing Subsequence

Return the length of the longest strictly increasing subsequence.

[Open on LeetCode](https://leetcode.com/problems/longest-increasing-subsequence/) · **Complexity:** O(n log n) time, O(n) space

```python
class Solution:
    def lengthOfLIS(self, nums: List[int]) -> int:
        tails = []
        for x in nums:
            i = bisect_left(tails, x)
            if i == len(tails):
                tails.append(x)
            else:
                tails[i] = x
        return len(tails)
```

### 32. Longest Valid Parentheses

Find the longest well-formed parentheses substring.

[Open on LeetCode](https://leetcode.com/problems/longest-valid-parentheses/) · **Complexity:** O(n) time, O(n) space

```python
class Solution:
    def longestValidParentheses(self, s: str) -> int:
        stack = [-1]
        best = 0
        for i, ch in enumerate(s):
            if ch == "(":
                stack.append(i)
            else:
                stack.pop()
                if not stack:
                    stack.append(i)
                else:
                    best = max(best, i - stack[-1])
        return best
```

### 152. Maximum Product Subarray

Return the largest product of a non-empty contiguous subarray.

[Open on LeetCode](https://leetcode.com/problems/maximum-product-subarray/) · **Complexity:** O(n) time, O(1) space

```python
class Solution:
    def maxProduct(self, nums: List[int]) -> int:
        low = high = best = nums[0]
        for x in nums[1:]:
            low, high = min(x, x * low, x * high), max(x, x * low, x * high)
            best = max(best, high)
        return best
```

### 416. Partition Equal Subset Sum

Determine whether the array can be split into two subsets with equal sums.

[Open on LeetCode](https://leetcode.com/problems/partition-equal-subset-sum/) · **Complexity:** O(n·sum) time, O(sum) space

```python
class Solution:
    def canPartition(self, nums: List[int]) -> bool:
        total = sum(nums)
        if total % 2:
            return False
        target = total // 2
        possible = {0}
        for x in nums:
            possible |= {v + x for v in possible if v + x <= target}
        return target in possible
```

### 62. Unique Paths

Count right/down paths across an `m × n` grid.

[Open on LeetCode](https://leetcode.com/problems/unique-paths/) · **Complexity:** O(mn) time, O(n) space

```python
class Solution:
    def uniquePaths(self, m: int, n: int) -> int:
        dp = [1] * n
        for _ in range(m - 1):
            for j in range(1, n):
                dp[j] += dp[j - 1]
        return dp[-1]
```

### 64. Minimum Path Sum

Find the minimum-sum right/down path through a nonnegative grid.

[Open on LeetCode](https://leetcode.com/problems/minimum-path-sum/) · **Complexity:** O(mn) time, O(n) space

```python
class Solution:
    def minPathSum(self, grid: List[List[int]]) -> int:
        dp = [inf] * len(grid[0])
        dp[0] = 0
        for row in grid:
            for j, x in enumerate(row):
                dp[j] = x + min(dp[j], dp[j - 1] if j else inf)
        return dp[-1]
```

### 5. Longest Palindromic Substring

Return the longest palindromic substring.

[Open on LeetCode](https://leetcode.com/problems/longest-palindromic-substring/) · **Complexity:** O(n²) time, O(1) space

```python
class Solution:
    def longestPalindrome(self, s: str) -> str:
        best = ""
        for center in range(2 * len(s) - 1):
            l = center // 2
            r = l + center % 2
            while l >= 0 and r < len(s) and s[l] == s[r]:
                if r - l + 1 > len(best):
                    best = s[l : r + 1]
                l -= 1
                r += 1
        return best
```

### 1143. Longest Common Subsequence

Return the LCS length of two strings.

[Open on LeetCode](https://leetcode.com/problems/longest-common-subsequence/) · **Complexity:** O(mn) time, O(n) space

```python
class Solution:
    def longestCommonSubsequence(self, a: str, b: str) -> int:
        dp = [0] * (len(b) + 1)
        for x in a:
            prev = 0
            for j, y in enumerate(b, 1):
                old = dp[j]
                dp[j] = prev + 1 if x == y else max(dp[j], dp[j - 1])
                prev = old
        return dp[-1]
```

### 72. Edit Distance

Return the minimum insertions, deletions, and replacements to transform one word into another.

[Open on LeetCode](https://leetcode.com/problems/edit-distance/) · **Complexity:** O(mn) time, O(n) space

```python
class Solution:
    def minDistance(self, a: str, b: str) -> int:
        dp = list(range(len(b) + 1))
        for i, x in enumerate(a, 1):
            new = [i] + [0] * len(b)
            for j, y in enumerate(b, 1):
                new[j] = dp[j - 1] if x == y else 1 + min(dp[j], new[j - 1], dp[j - 1])
            dp = new
        return dp[-1]
```

### 121. Best Time to Buy and Sell Stock

Maximize profit from one buy followed by one sell.

[Open on LeetCode](https://leetcode.com/problems/best-time-to-buy-and-sell-stock/) · **Complexity:** O(n) time, O(1) space

```python
class Solution:
    def maxProfit(self, prices: List[int]) -> int:
        low, best = inf, 0
        for p in prices:
            low = min(low, p)
            best = max(best, p - low)
        return best
```

### 55. Jump Game

Determine whether the last index is reachable.

[Open on LeetCode](https://leetcode.com/problems/jump-game/) · **Complexity:** O(n) time, O(1) space

```python
class Solution:
    def canJump(self, nums: List[int]) -> bool:
        far = 0
        for i, x in enumerate(nums):
            if i > far:
                return False
            far = max(far, i + x)
        return True
```

### 45. Jump Game II

Return the fewest jumps needed to reach the last index.

[Open on LeetCode](https://leetcode.com/problems/jump-game-ii/) · **Complexity:** O(n) time, O(1) space

```python
class Solution:
    def jump(self, nums: List[int]) -> int:
        jumps = end = far = 0
        for i in range(len(nums) - 1):
            far = max(far, i + nums[i])
            if i == end:
                jumps += 1
                end = far
        return jumps
```

### 763. Partition Labels

Partition a string so each letter occurs in at most one part; return part sizes.

[Open on LeetCode](https://leetcode.com/problems/partition-labels/) · **Complexity:** O(n) time, O(k) space

```python
class Solution:
    def partitionLabels(self, s: str) -> List[int]:
        last = {ch: i for i, ch in enumerate(s)}
        start = end = 0
        out = []
        for i, ch in enumerate(s):
            end = max(end, last[ch])
            if i == end:
                out.append(end - start + 1)
                start = i + 1
        return out
```
